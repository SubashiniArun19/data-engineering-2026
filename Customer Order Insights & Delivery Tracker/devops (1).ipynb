```python
process_code = '''
import pandas as pd
print("Processing Customer Orders...")
df = pd.read_csv("orders.csv")
delay_summary = df.groupby( "customer_id" ).size()
print(delay_summary)
with open( "delay_summary_log.txt", "w" ) as log_file:
  log_file.write( str(delay_summary) )
  print("Delay summary log created")'''
with open( "process_orders.py", "w" ) as file:
  file.write(process_code)
print("process_orders.py created")
```

```python
pipeline_code = '''
print("Starting Customer Delivery Pipeline...")
exec(open("process_orders.py").read())
print("Pipeline Execution Completed Successfully") '''
with open( "run_pipeline.py", "w" ) as file:
  file.write(pipeline_code)
print("run_pipeline.py created")
```
