Parquet is a columnar file format used only for efficient data storage and reads.
It does not support updates, deletes, or ACID transactions.
Delta Lake is built on top of Parquet and adds a transaction log for data management.
It supports updates, deletes, time travel, and ensures data consistency with ACID operations.

```python
import dlt
from pyspark.sql.functions import *

@dlt.table
def bronze_orders():
    return spark.createDataFrame(daily_orders_data, daily_orders_columns)
```

```python
@dlt.table
def silver_orders():
    return dlt.read("bronze_orders") \
        .withColumn("order_date", to_date("order_date", "yyyy-MM-dd"))
```

```python
@dlt.table
def silver_orders_enriched():
    products = spark.table("products_delta")  # assume already saved

    return dlt.read("silver_orders") \
        .join(products, "product_id") \
        .withColumn("total_revenue", col("quantity") * col("price"))
```

```python
@dlt.table
def silver_clean_orders():
    return dlt.read("silver_orders_enriched") \
        .filter(col("quantity") > 0) \
        .filter(col("total_revenue") > 0)
```

```python
@dlt.table
def gold_city_revenue():
    return dlt.read("silver_clean_orders") \
        .groupBy("warehouse_city") \
        .agg(sum("total_revenue").alias("city_revenue"))
```

```python
@dlt.table
def gold_category_revenue():
    return dlt.read("silver_clean_orders") \
        .groupBy("category") \
        .agg(sum("total_revenue").alias("category_revenue"))
```
