```python
process_code = """
import pandas as pd
import numpy as np
data = {"order_id": [1,2,3,4],"supplier_id": [101,102,101,103],"delivery_date": ["2026-05-01","2026-05-10","2026-05-14","2026-04-28"]}
df = pd.DataFrame(data)
df['delivery_date'] = pd.to_datetime(df['delivery_date'])
df['delay_days'] = (pd.Timestamp.today() -df['delivery_date']).dt.days
df['is_delayed'] = np.where(df['delay_days'] > 0,1,0)
print(df)
df.to_csv("processed_orders.csv",index=False)
print("Processing Completed")
"""

with open("process_orders.py", "w") as file:
    file.write(process_code)

print("process_orders.py created successfully")
```

```python
print("Starting Supply Chain Pipeline...")
exec(open("process_orders.py").read())
print("Pipeline Execution Completed")
```
